```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import (train_test_split, StratifiedKFold, KFold,
cross_val_score, learning_curve, GridSearchCV)
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import (LogisticRegression, LinearRegression,
Ridge, Lasso, ElasticNet)
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, r2_score, mean_squared_error
```

```python
pd.set_option("display.width", 120)
pd.set_option("display.max_columns", 20)
```

```python
df = pd.read_csv("/diabetes.csv")
hidden_missing = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
df[hidden_missing] = df[hidden_missing].replace(0, np.nan)
```

```python
X = df.drop(columns="Outcome")
y = df["Outcome"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.20, stratify=y, random_state=42)
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
print("Training set:", X_train.shape, " Test set:", X_test.shape)
```

```python
depths = list(range(1, 21))
train_acc, cv_acc = [], []
```

```python
for d in depths:
    tree = DecisionTreeClassifier(max_depth=d, random_state=42)
    pipe = Pipeline([("imputer", SimpleImputer(strategy="median")),
                    ("model", tree)])
    pipe.fit(X_train, y_train)
    train_acc.append(accuracy_score(y_train, pipe.predict(X_train)))
    cv_acc.append(cross_val_score(pipe, X_train, y_train, cv=cv).mean())
```

```python
curve = pd.DataFrame({"max_depth": depths, "train_acc": train_acc,
"cv_acc": cv_acc})
curve["gap"] = curve["train_acc"] - curve["cv_acc"]
print(curve.round(3).to_string(index=False))
```

```python
# Cost-complexity pruning: a larger ccp_alpha removes more branches.
imputer = SimpleImputer(strategy="median")
X_train_imp = imputer.fit_transform(X_train)

```

```python
full_tree = DecisionTreeClassifier(random_state=42)
path = full_tree.cost_complexity_pruning_path(X_train_imp, y_train)
alphas = path.ccp_alphas[:-1]   # the last alpha would leave a single node
alphas = alphas[::4]            # every 4th value, to keep the loop short
print("Number of alpha values tried:", len(alphas))

```

```python

pr_train, pr_cv, pr_leaves = [], [], []
for a in alphas:
    tree = DecisionTreeClassifier(ccp_alpha=a, random_state=42)
    pipe = Pipeline([("imputer", SimpleImputer(strategy="median")),
                     ("model", tree)])
    pipe.fit(X_train, y_train)
    pr_train.append(accuracy_score(y_train, pipe.predict(X_train)))
    pr_cv.append(cross_val_score(pipe, X_train, y_train, cv=cv).mean())
    pr_leaves.append(pipe.named_steps["model"].get_n_leaves())

```

```python
best_alpha = alphas[int(np.argmax(pr_cv))]
print("Best ccp_alpha:", round(best_alpha, 5),
      " CV accuracy:", round(max(pr_cv), 3),
      " leaves:", pr_leaves[int(np.argmax(pr_cv))])

```

```python
plt.figure(figsize=(8, 4.5))
plt.plot(alphas, pr_train, marker="o", label="Training accuracy")
plt.plot(alphas, pr_cv, marker="s", label="Cross-validation accuracy")
plt.axvline(best_alpha, color="grey", linestyle="--")
plt.xlabel("ccp_alpha (regularization strength)")
plt.ylabel("Accuracy")
plt.title("Pruning a decision tree")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# Degree-3 polynomial features turn 8 inputs into 164 inputs, which makes the
# logistic model flexible enough to overfit. C controls the L2 penalty:
# small C = strong regularization, large C = almost no regularization.
C_values = np.logspace(-4, 3, 8)
lr_train, lr_cv = [], []
for C in C_values:
    pipe = Pipeline([("imputer", SimpleImputer(strategy="median")),
                     ("scaler", StandardScaler()),
                     ("poly", PolynomialFeatures(degree=3, include_bias=False)),
                     ("scaler2", StandardScaler()),
                     ("model", LogisticRegression(C=C, max_iter=5000))])
    pipe.fit(X_train, y_train)
    lr_train.append(accuracy_score(y_train, pipe.predict(X_train)))
    lr_cv.append(cross_val_score(pipe, X_train, y_train, cv=cv).mean())

```

```python
lr_table = pd.DataFrame({"C": C_values, "train_acc": lr_train, "cv_acc": lr_cv})
lr_table["gap"] = lr_table["train_acc"] - lr_table["cv_acc"]
cols = ["train_acc", "cv_acc", "gap"]
lr_table[cols] = lr_table[cols].round(3)
print(lr_table.to_string(index=False, formatters={"C": "{:g}".format}))

plt.figure(figsize=(8, 4.5))
plt.semilogx(C_values, lr_train, marker="o", label="Training accuracy")
plt.semilogx(C_values, lr_cv, marker="s", label="Cross-validation accuracy")
plt.xlabel("C  (smaller = stronger regularization)")
plt.ylabel("Accuracy")
plt.title("Regularized logistic regression with 164 polynomial features")
plt.legend()
plt.grid(alpha=0.3)
plt.show()


```

```python
ins = pd.read_csv("/insurance.csv")
print(ins.head())
print()
print(ins.dtypes)
print()
```

```python
# Convert text columns into 0/1 columns
ins = pd.get_dummies(ins, columns=["sex", "smoker", "region"],
                     drop_first=True, dtype=int)
print(ins.columns.tolist())

```

```python
Xr = ins.drop(columns="charges")
yr = ins["charges"]
Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    Xr, yr, test_size=0.20, random_state=42)

cv_reg = KFold(n_splits=5, shuffle=True, random_state=42)
print()
print("Training:", Xr_train.shape, " Test:", Xr_test.shape)
```

```python
def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

def evaluate(name, pipe):
    pipe.fit(Xr_train, yr_train)
    cv_r2 = cross_val_score(pipe, Xr_train, yr_train, cv=cv_reg,
                            scoring="r2").mean()
    return {"model": name,
            "train_R2": r2_score(yr_train, pipe.predict(Xr_train)),
            "cv_R2": cv_r2,
            "test_R2": r2_score(yr_test, pipe.predict(Xr_test)),
            "test_RMSE": rmse(yr_test, pipe.predict(Xr_test))}

def poly_pipe(degree, model):
    return Pipeline([("scaler", StandardScaler()),
                     ("poly", PolynomialFeatures(degree=degree,
                                                 include_bias=False)),
                     ("scaler2", StandardScaler()),
                     ("model", model)])

results = [
    evaluate("Linear, degree 1 (too simple)", poly_pipe(1, LinearRegression())),
    evaluate("Linear, degree 2", poly_pipe(2, LinearRegression())),
    evaluate("Linear, degree 3", poly_pipe(3, LinearRegression())),
    evaluate("Linear, degree 4 (too complex)", poly_pipe(4, LinearRegression())),
]
print(pd.DataFrame(results).set_index("model").round(3))
```

```python
alphas_ridge = np.logspace(-2, 4, 13)
ridge_rows = []
for a in alphas_ridge:
    pipe = poly_pipe(4, Ridge(alpha=a))
    pipe.fit(Xr_train, yr_train)
    ridge_rows.append({"alpha": a,
                       "train_R2": r2_score(yr_train, pipe.predict(Xr_train)),
                       "cv_R2": cross_val_score(pipe, Xr_train, yr_train,
                                                cv=cv_reg, scoring="r2").mean(),
                       "test_R2": r2_score(yr_test, pipe.predict(Xr_test))})

ridge_table = pd.DataFrame(ridge_rows)
print(ridge_table.round(3).to_string(index=False))

plt.figure(figsize=(8, 4.5))
plt.semilogx(ridge_table["alpha"], ridge_table["train_R2"],
             marker="o", label="Training R2")
plt.semilogx(ridge_table["alpha"], ridge_table["cv_R2"],
             marker="s", label="Cross-validation R2")
plt.xlabel("alpha (larger = stronger regularization)")
plt.ylabel("R-squared")
plt.title("Ridge regression on degree-4 polynomial features")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

```

```python
# Very small alpha values make Lasso slow to converge on this many features,
# so the search starts at alpha = 10.
alphas_lasso = np.logspace(1, 3.5, 8)
lasso_rows = []
for a in alphas_lasso:
    pipe = poly_pipe(4, Lasso(alpha=a, max_iter=20000))
    pipe.fit(Xr_train, yr_train)
    coefs = pipe.named_steps["model"].coef_
    lasso_rows.append({"alpha": a,
                       "non_zero_coefs": int(np.sum(coefs != 0)),
                       "train_R2": r2_score(yr_train, pipe.predict(Xr_train)),
                       "cv_R2": cross_val_score(pipe, Xr_train, yr_train,
                                                cv=cv_reg, scoring="r2").mean(),
                       "test_R2": r2_score(yr_test, pipe.predict(Xr_test))})

lasso_table = pd.DataFrame(lasso_rows)
n_features = poly_pipe(4, Ridge()).fit(Xr_train, yr_train)
n_features = n_features.named_steps["model"].coef_.shape[0]
print("Total number of features available:", n_features)
print()
print(lasso_table.round(3).to_string(index=False))
```

```python
import warnings
from sklearn.exceptions import ConvergenceWarning
# Lasso/ElasticNet with small alpha sometimes stop before fully converging;
# the results are still usable here, so we hide the repeated warning messages.
warnings.filterwarnings("ignore", category=ConvergenceWarning)

param_grids = {
    "Ridge": (poly_pipe(4, Ridge()),
              {"model__alpha": np.logspace(-2, 4, 13)}),
    "Lasso": (poly_pipe(4, Lasso(max_iter=20000)),
              {"model__alpha": np.logspace(1, 3.5, 8)}),
    "ElasticNet": (poly_pipe(4, ElasticNet(max_iter=20000)),
                   {"model__alpha": np.logspace(-0.5, 2.5, 7),
                    "model__l1_ratio": [0.2, 0.5, 0.8]}),
}

final = [evaluate("No regularization (degree 4)", poly_pipe(4, LinearRegression()))]
for name, (pipe, grid) in param_grids.items():
    search = GridSearchCV(pipe, grid, cv=cv_reg, scoring="r2", n_jobs=-1)
    search.fit(Xr_train, yr_train)
    print(f"{name:10s} best settings: {search.best_params_}")
    best = search.best_estimator_
    final.append({"model": f"{name} (tuned)",
                  "train_R2": r2_score(yr_train, best.predict(Xr_train)),
                  "cv_R2": search.best_score_,
                  "test_R2": r2_score(yr_test, best.predict(Xr_test)),
                  "test_RMSE": rmse(yr_test, best.predict(Xr_test))})

print()
print(pd.DataFrame(final).set_index("model").round(3))
```

```python
# Organize and display the final comparison results of the regression models in a table
final_df = pd.DataFrame(final).set_index("model")
final_df["gap"] = final_df["train_R2"] - final_df["test_R2"]
print("=== Regression Model Performance Comparison ===")
print(final_df[["train_R2", "cv_R2", "test_R2", "test_RMSE", "gap"]].round(3))
```
